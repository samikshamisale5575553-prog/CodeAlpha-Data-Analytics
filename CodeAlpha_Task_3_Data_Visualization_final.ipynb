## 1. Importing Required Libraries

Pandas is used for data handling, NumPy for numerical operations, Matplotlib for plotting, and Seaborn for statistical visualizations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
print("All libraries imported successfully!")
```

## 2. Loading the Dataset

The dataset is loaded from the publicly available CSV file. The semicolon separator (`;`) is required for this dataset.

```python
url = "https://raw.githubusercontent.com/MakroMB/student-performance-analysis/main/data/student-por.csv"

df = pd.read_csv(url, sep=";")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

## 3. Dataset Preview

The first five rows, column names, and data types are displayed to understand the structure of the dataset.

```python
display(df.head())

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()
```

## 4. Data Visualization

The following sections contain distinct visualizations. Each chart includes a title and labelled axes to make the results easier to interpret.

### 4.1 Distribution of Final Grades (Histogram)

A histogram shows how frequently different final-grade values occur.

```python
plt.figure(figsize=(8, 5))
plt.hist(df["G3"], bins=10, edgecolor="black")
plt.title("Distribution of Final Grades (G3)")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

### 4.2 Average Final Grade by Study Time

This bar chart compares the average final grade across study-time categories. In this dataset, study-time categories are coded from 1 to 4.

```python
avg_grade_study = df.groupby("studytime")["G3"].mean()

plt.figure(figsize=(8, 5))
sns.barplot(x=avg_grade_study.index, y=avg_grade_study.values, errorbar=None)
plt.title("Average Final Grade by Study Time")
plt.xlabel("Study Time Category")
plt.ylabel("Average Final Grade")
plt.tight_layout()
plt.show()
```

### 4.3 Study Time and Final Grades (Box Plot)

A box plot compares the spread and median of final grades across study-time categories.

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="studytime", y="G3")
plt.title("Study Time vs Final Grades")
plt.xlabel("Study Time Category")
plt.ylabel("Final Grade (G3)")
plt.tight_layout()
plt.show()
```

### 4.4 Distribution of Final Grades (Box Plot)

```python
plt.figure(figsize=(8, 5))
sns.boxplot(y=df["G3"])
plt.title("Box Plot of Final Grades (G3)")
plt.ylabel("Final Grade")
plt.tight_layout()
plt.show()
```

### 4.5 Average Final Grades by Previous Class Failures

### 4.7 Correlation Heatmap

The heatmap displays Pearson correlations among numerical variables. Values closer to 1 indicate positive linear association, values closer to -1 indicate negative linear association, and values near 0 indicate weak linear association.

```python
plt.figure(figsize=(14, 10))
correlation_matrix = df.corr(numeric_only=True)

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    annot=True,
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Student Performance Variables")
plt.tight_layout()
plt.show()
```

## 5. Visual Summary Dashboard

This dashboard brings together four complementary views of the dataset: the distribution of final grades, average grades by study-time category, average grades by previous-failure category, and the relationship between absences and final grades. It provides a quick overview before the detailed interpretation.


```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Overall distribution of final grades
sns.histplot(data=df, x="G3", bins=10, kde=True, ax=axes[0, 0])
axes[0, 0].set_title("Distribution of Final Grades")
axes[0, 0].set_xlabel("Final Grade (G3)")
axes[0, 0].set_ylabel("Number of Students")

# 2. Average final grade by study-time category
study_means = df.groupby("studytime", observed=False)["G3"].mean()
sns.barplot(x=study_means.index, y=study_means.values, errorbar=None, ax=axes[0, 1])
axes[0, 1].set_title("Average Grade by Study Time")
axes[0, 1].set_xlabel("Study-Time Category")
axes[0, 1].set_ylabel("Average Final Grade")

# 3. Average final grade by previous-failure category
failure_means = df.groupby("failures", observed=False)["G3"].mean()
sns.barplot(x=failure_means.index, y=failure_means.values, errorbar=None, ax=axes[1, 0])
axes[1, 0].set_title("Average Grade by Previous Failures")
axes[1, 0].set_xlabel("Number of Previous Failures")
axes[1, 0].set_ylabel("Average Final Grade")

# 4. Absences compared with final grades
sns.scatterplot(data=df, x="absences", y="G3", alpha=0.6, ax=axes[1, 1])
axes[1, 1].set_title("Absences and Final Grades")
axes[1, 1].set_xlabel("Number of Absences")
axes[1, 1].set_ylabel("Final Grade (G3)")

fig.suptitle("Student Performance: Visual Summary", fontsize=16, y=1.02)
plt.tight_layout()
plt.show()
```

## 6. Key Findings from Visualizations

Use the charts above to review the patterns in the dataset. The following code calculates selected group averages so that the written findings can be checked against the actual dataset output.

```python
print("Average final grade by study time:")
display(df.groupby("studytime")["G3"].mean().round(2))

print("\nAverage final grade by previous failures:")
display(df.groupby("failures")["G3"].mean().round(2))

print("\nAverage final grade by mother's education:")
display(df.groupby("Medu")["G3"].mean().round(2))

print("\nAverage final grade by father's education:")
display(df.groupby("Fedu")["G3"].mean().round(2))

print("\nAverage final grade by going-out frequency:")
display(df.groupby("goout")["G3"].mean().round(2))

print("\nCorrelation of numerical variables with G3:")
display(df.corr(numeric_only=True)["G3"].sort_values(ascending=False).round(3))
```

## 7. Data Storytelling and Interpretation

A useful data story connects the visual evidence to a clear message. The code below calculates selected statistics directly from the loaded dataset and creates short interpretations from those results. This avoids hard-coding conclusions that may not match the data.


```python
# Calculate dataset-specific summaries for a reproducible data story
mean_grade = df["G3"].mean()
median_grade = df["G3"].median()
zero_grade_share = (df["G3"] == 0).mean() * 100

study_means = df.groupby("studytime")["G3"].mean()
failure_means = df.groupby("failures")["G3"].mean()
absence_corr = df[["absences", "G3"]].corr().iloc[0, 1]
g1_g3_corr = df[["G1", "G3"]].corr().iloc[0, 1]

best_study_category = study_means.idxmax()
lowest_failure_category = failure_means.idxmin()

print("DATA STORY: STUDENT PERFORMANCE")
print("-" * 40)
print(f"The average final grade is {mean_grade:.2f}, while the median is {median_grade:.2f}.")
print(f"Students in study-time category {best_study_category} have the highest average final grade ({study_means.max():.2f}) among the study-time categories.")
print(f"The previous-failure category with the lowest average final grade is {lowest_failure_category} ({failure_means.min():.2f}).")
print(f"The correlation between absences and final grades is {absence_corr:.2f}.")
print(f"The correlation between first-period grades (G1) and final grades (G3) is {g1_g3_corr:.2f}.")
print(f"{zero_grade_share:.1f}% of records have a final grade of zero.")

print("\\nSTORY IN ONE PARAGRAPH")
print(
    "The visual analysis describes how final grades are distributed and how average outcomes "
    "vary across study-time and previous-failure categories. The correlation values summarize "
    "linear associations between absences, earlier grades, and final grades. These results can "
    "help identify patterns worth investigating, but they do not show that any one factor caused "
    "a student's result."
)
```

### Evidence-Based Recommendations

The following are cautious suggestions for further investigation based on the kinds of patterns shown in this notebook. They are not causal claims or one-size-fits-all conclusions.

- **Monitor academic progress:** Compare earlier-period grades (`G1` and `G2`) with final grades to identify students who may benefit from timely academic support.
- **Explore study habits:** Review study-time categories alongside grade distributions rather than relying only on category averages.
- **Understand attendance patterns:** Examine absences together with grades and consider additional context before drawing conclusions.
- **Investigate previous academic difficulties:** Use the previous-failure comparisons to identify areas where additional learning support may be worth evaluating.
- **Use multiple indicators:** Avoid making decisions from a single chart; consider grade history, attendance, study patterns, and relevant student context together.

Further analysis with additional information would be needed before making policy or individual-level decisions.
